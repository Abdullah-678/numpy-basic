```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
sales_data=np.array([
  [1,15000,18000,22000,25000],
  [2,12000,14000,16000,19000],
  [3,20000,23000,26000,30000],
  [4,18000,21000,24000,27000],
  [5,15000,18500,20000,23000],
])
```

```python
print("\n Sales data Shape",sales_data.shape)
print("\n Sample data for 1st 3 :",sales_data[:3])
```

```python
## total sales 
print(np.sum(sales_data,axis=0))

##yearly sales
print(np.sum(sales_data[:,1:],axis=0))


##Minimum sales
print(np.min(sales_data[:,1:],axis=1))

##maximum sales
print(np.max(sales_data[:,1:],axis=0))


##Average sales per restaurant
print(np.mean(sales_data[:,1:],axis=1))
```

```python
cumsum=np.cumsum(sales_data[:,1:],axis=1)
print(cumsum)

plt.figure(figsize=(10,6))
plt.plot(np.mean(cumsum,axis=0))
plt.title("Average cumutative sales accross all restaurant")
plt.xlabel("Years")
plt.ylabel("sales")
plt.grid(True)
plt.show()
```

```python
vector1=np.array([1,2,3,4,5])
vector2=np.array([6,7,8,9,10])
print("vector additon",vector1+vector2)
print("\n multiplication vector",vector1*vector2)
print("\n Dot product",np.dot(vector1,vector2))
angle=np.arccos(np.dot(vector1,vector2)/(np.linalg.norm(vector1)*np.linalg.norm(vector2)))
print(angle)
```
